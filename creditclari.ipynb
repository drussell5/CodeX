```python
h5ert7import gradio as gr
import pytesseract
from PIL import Image
import openai
import os
# Set your OpenAI API Key
openai.api_key = os.environ.get("OPENAI_API_KEY")
def analyze_credit_report(image):
  # Step 1: OCR to extract text
  extracted_text = pytesseract.image_to_string(image)
  # Step 2: ASK GPT to find disputable items
  prompt = f"Analyze this credit report and find any disputable items: {extracted_text}"
  response = openai.Completion.create(
    engine="text-davinci-002",
    prompt=prompt,
    max_tokens=1024,
    n=1,
    stop=None,
    temperature=0.5,
  )
  ai_output = response.choices[0].text.strip()
  return ai_output
# Gradio UI
interface = gr.Interface(
      fn=analyze_credit_report,
      inputs="image",
      outputs="text", # This line was incorrectly indented
      title="Credit Report Analyzer",
      description="Upload a credit report image and get insights from GPT-3.",
  )
interface.launch()
```
